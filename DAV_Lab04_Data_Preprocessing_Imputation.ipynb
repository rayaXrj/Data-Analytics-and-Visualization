```python
import numpy as np
import pandas as pd
```

```python
data = {
    "Customer_ID": ["C101","C102","C103","C104","C105","C106","C107","C108","C109","C110"],
    "City": ["Bengaluru","Mysuru","Hubballi","Bengaluru","Mangaluru",
             "Mysuru","Hubballi","Bengaluru","Mangaluru","Mysuru"],
    "Age": [25, np.nan, 31, 28, 42, 35, np.nan, 29, 38, 27],
    "Income": [45000, 38000, np.nan, 52000, 47000, 41000, 36000, np.nan, 55000, 40000],
    "Membership": ["Gold","Silver","Gold",np.nan,"Silver",
                   "Gold","Bronze","Gold","Silver","Bronze"],
    "Purchase": [3200,2100,4100,3500,2800,3900,1800,4600,5200,2400]
}

df = pd.DataFrame(data)
df
```

```python
print("Shape:",df.shape)
print("\nData types:")
print(df.dtypes)
print("\nFirst five rows")
display(df.head())
```

```python
missing_count = df.isnull().sum()
missing_count
```

```python
missing_percentage = (df.isnull().sum()/len(df)) *100
missing_percentage
```

```python
age_mean = df["Age"].mean()
print("Mean age:",age_mean)
df_mean=df.copy()
df_mean["Age"].fillna(age_mean)
df_mean
```

```python
age_mean = df["Age"].mean()
print("Mean age:",age_mean)
df_mean=df.copy()
print(df_mean)
df_mean["Age"] = df_mean["Age"].fillna(age_mean)
print("\nAge after mean imputation:")
print(df_mean["Age"])
```

```python
income_median = df["Income"].median()
print("Median income:",income_median)
df_median=df.copy()
print(df_median)
df_median["Income"] = df_median["Income"].fillna(income_median)
print("\nIncome after median imputation:")
print(df_median["Income"])
```

```python
values = pd.Series([20,25,30,35,100])
print("Mean:", values.mean)
print("Median:", values.median)
print("Mode:", values.mode)
```

```python
weather = pd.DataFrame({
    "Day":["Monday","Tuesday","Wednesday","Thursday","Friday","Saturday","Sunday"],
    "Temperature":[30,30,np.nan,32,np.nan,31,33]

})
weather
```

```python
weather_ffill = weather.copy()
weather_ffill["Temperature"] = weather_ffill["Temperature"].fillna(method="ffill")
weather_ffill
```

```python
copydf = df.copy()
print(copydf)
copydf["Age"] = copydf["Age"].fillna(copydf["Age"].mean())
print(copydf["Age"])
copydf["Income"] = copydf["Income"].fillna(copydf["Income"].median())
print(copydf[["Age","Income","Membership"]].isnull().sum())
```

```python
membership_mode = copydf["Membership"].mode()[0]
print("Most frequent Membership:",membership_mode)
copydf["Membership"] =copydf["Membership"].fillna(membership_mode)

print("\nMissing Membership values:",copydf["Membership"].isnull().sum())
```

```python
encoded_df = pd.get_dummies( copydf,
                            columns=["City","Membership"],
                            dtype=int
                            )
encoded_df
```

```python
print(encoded_df.columns.tolist())
```

```python

print("Missing values: ")
print(encoded_df.isnull().sum())

print("\n Data types: ")
print(encoded_df.dtypes)

print("\n Final shape: ",encoded_df.shape)
```
