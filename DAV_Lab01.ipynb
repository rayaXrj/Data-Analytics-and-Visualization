```python
import numpy as np
import pandas as pd
print("Numpy Version:",np.__version__)
print("Pandas Version:",pd.__version__)
```

```python
marks=np.array([89,67,55,45,87])
print("marks",marks)
print("type of marks",type(marks))
```

```python
marks=np.array([89,67,55,45,87])
print("Original",marks)
print("Add 5",marks + 5)
print("multiply by 2",marks*2)
print("Divided by 2:",marks/2)
print("subtract by 2:",marks - 2)
print("square:",marks**2)
```

```python
print("sum:",marks.sum())
print("mean:",marks.mean())
print("Maximum:",marks.max())
print("Minimum:",marks.min())
```

```python
print("index 1 to 3",marks[1:4])
print("first three",marks[:3])
print("from index 0 to 3",marks[:4])
```

```python
student_marks=np.array([
    [78,67,89],
    [89,88,76],
    [90,95,90]])
print(student_marks)
print("shape of the array:",student_marks.shape)
```

```python
print("Mean of all values:",student_marks.mean())
print("Column-wise Mean:",student_marks.mean(axis=0))
print("Row-wise mean",student_marks.mean(axis=1))
```

```python
data = {
    "Name": ["Ravana","Rahul","Ticket","Raya"],
    "Branch":["CSE","AIML","CSE","AIML"],
    "Marks":[85,99,78,76]
}
df = pd.DataFrame(data)
print(df)
```

```python
print("Shape:",df.shape)
print("\nColumn:")
print(df.columns)
```

```python
print("First Row:")
print(df.iloc[0])
print("\nSecond Row:")
print(df.iloc[1])
print("\nFirst three Rows:")
print(df.iloc[0:3])
print("First two rows and first two columns:")
print(df.iloc[0:2, 0:2])
```
